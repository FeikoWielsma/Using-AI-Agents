# Direct an agent and build working memory

- Give an agent a bounded task using a realistic fictional document pack
- Ask it to help create a brief, memory and reusable procedures in Markdown
- Review its evidence and file changes, then resume in a fresh conversation

Run the setup and interface cells below. Everything you change during the exercise
is entered in the interface. No Python editing is needed. The notebook keeps its
original link so bookmarks from this course continue working.

In Colab, add `COURSE_API_KEY` under the key icon and enable Notebook access.
Use the key supplied by the trainer. Made up, not from work: use the fictional
course records and do not paste employer, customer or personal information.


```python
#@title Open the course workspace
import os, sys, subprocess, json, urllib.request, importlib.util
from pathlib import Path
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'openai==2.41.1', 'ipywidgets>=8,<9'], check=True)
BASE_URL = 'https://raw.githubusercontent.com/FeikoWielsma/Using-AI-Agents/main/'
def course_file(name):
    local_root = os.environ.get('COURSE_LOCAL_ROOT')
    if local_root:
        return Path(local_root) / name
    path = Path('course-assets') / name
    path.parent.mkdir(parents=True, exist_ok=True)
    urllib.request.urlretrieve(BASE_URL + name, path)
    return path
KEY = os.environ.get('COURSE_API_KEY')
if not KEY:
    try:
        from google.colab import userdata
        KEY = userdata.get('COURSE_API_KEY')
    except Exception:
        KEY = None
from openai import OpenAI
MODEL = 'qwen3.8-max'
EXTRA = {'enable_thinking': False}
client = OpenAI(api_key=KEY or 'course-key-not-configured',
    base_url='https://token-plan.ap-southeast-1.maas.aliyuncs.com/compatible-mode/v1',
    timeout=35.0, max_retries=0)
spec=importlib.util.spec_from_file_location('course_workbench',course_file('notebooks/course_workbench.py'))
workbench=importlib.util.module_from_spec(spec)
spec.loader.exec_module(workbench)
print('Course interface ready.' if KEY else 'Add COURSE_API_KEY in Colab Secrets, enable notebook access, and rerun this setup. Source browsing remains available.')
```

```python
#@title Show the workplace agent
PACKS=json.loads(course_file('data/case-packs.json').read_text(encoding='utf-8'))
session=workbench.show_workbench(PACKS,client,MODEL,EXTRA)
```

## 1. Inspect and plan

1. Choose Launch readiness. Read `sources/case-overview.md` through the file selector.
2. Run the suggested Inspect and plan prompt. Watch the tool activity.
3. Identify one source you would check yourself. Read it in full in the file viewer.
4. Improve the brief: request source references, separate forecasts from decisions,
   and prohibit sending a stakeholder message without review.
5. Ask the agent to propose `BRIEF.md`. Select Review changes, read the proposal,
   and apply it only when it captures your intended task.

Does the brief say what completion means? Does it distinguish a draft from a sent
message? The agent proposing a document has not yet saved it.

## 2. Produce and check a result

1. Choose Do the work and run it using the saved brief.
2. Read the cited sources for the proposed result. Inspect G-01 and E-001.
3. Check the current checklist version, dates and decision authority.
4. Ask for a correction if an assumption is stated as a fact. Review and apply
   `RESULT.md` when its evidence and limitations are clear.

The executive summary and supporting records can disagree. The useful result
identifies the disagreement and its owner; it does not silently select a convenient
statement. The current records do not supply a final launch approval.

## 3. Ask the agent to maintain Markdown

1. Choose Capture useful memory. Review `MEMORY.md` for dated facts, source links,
   unresolved questions and the next step. Remove unnecessary chat history.
2. Choose Extract a procedure. Check that `PROCEDURE.md` describes reusable steps
   and checks, rather than treating this case's dates as permanent rules.
3. Ask the agent to improve `README.md` so another conversation knows where to start.
4. Review and apply the selected changes. Inspect the files in the file viewer.

`DECISIONS.md` records decisions you actually made and their reasons. A proposal
by the agent is not your decision. A plain Markdown file has no automatic memory
power; this workbench gives the agent file tools and a starting instruction.

## 4. Resume and revise

1. Select Fresh conversation, then run Resume from files.
2. Check what it recovered from the applied files. Unapplied proposals are not memory.
3. Select Release later evidence. Two new source records become available.
4. Run Incorporate new evidence. Review the changes to the memory and result.
5. Check which condition changed and which conditions still remain open.

A new source can supersede a fact without erasing the historical decision record.
The update is evidence, not an instruction to approve the launch.

## 5. Apply it to a professional case

1. Choose the quality, account, programme or banking-app case.
2. Inspect the available records and adapt the proposed task to a useful deliverable.
3. Repeat the brief, result, memory and procedure cycle. Keep the same checks.
4. Ask for `CHECKLIST.md` containing three acceptance checks and one challenging input.
5. Ask for `PROJECT-BRIEF.md` describing source access, authority, checks, human
   decisions and costs you would measure before a pilot.

## 6. Take the workspace with you

1. Apply the changes you want to keep. Choose Download workspace ZIP.
2. Keep the ZIP outside Colab. It contains source records and applied Markdown files.
3. In a fresh runtime, open this notebook, choose the matching case, and Restore ZIP.
4. Run Resume from files and check that the agreed brief and current memory survived.

Colab runtime files are temporary. Saving a notebook to Drive does not save its
runtime files. The ZIP excludes conversation history and unapplied proposals.
Restoring replaces this case's working documents with the exported version.

## If something does not work

- Missing key: enable access to `COURSE_API_KEY`, then rerun setup and the interface
- Request failed: inspect the visible status; retry once or use the supplied source
  documents and the manual fallback in the handout
- Turn limit: narrow the request, ask for a partial result, then continue
- No files proposed: explicitly ask it to propose the named Markdown document
- A proposal changed meanwhile: discard it and request an update against current files
- Controls absent: reconnect Colab and rerun the two interface cells
