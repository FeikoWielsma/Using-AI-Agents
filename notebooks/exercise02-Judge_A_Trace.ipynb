# Judge an agent's work

- Compare the result with the task, tools and complete evidence
- Separate incomplete work, missing evidence and a specification problem
- Compare your judgment with a model using the same review standard

The twelve stored runs are selected fault scenarios. Some intentionally lack tools,
some have contradictory instructions, and one has a broken search tool. They are
not a benchmark of normal reliability. Start with traces 2, 5, 7 and 8.

The original instructions are reconstructed from the generator and labelled as
such. The captured answer and tool results are unchanged. Both you and the model
judge receive the same complete record. No answer is cut off for display.

Run the two cells below. Use the interface to record a verdict and reason; no
Python editing is needed. The same Colab course key is used here.


```python
#@title Open the course workspace
import os, sys, subprocess, json, urllib.request, importlib.util
from pathlib import Path
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'openai==2.41.1', 'ipywidgets>=8,<9'], check=True)
BASE_URL = 'https://raw.githubusercontent.com/FeikoWielsma/Using-AI-Agents/main/'
def course_file(name):
    local_root = os.environ.get('COURSE_LOCAL_ROOT')
    if local_root:
        return Path(local_root) / name
    path = Path('course-assets') / name
    path.parent.mkdir(parents=True, exist_ok=True)
    urllib.request.urlretrieve(BASE_URL + name, path)
    return path
KEY = os.environ.get('COURSE_API_KEY')
if not KEY:
    try:
        from google.colab import userdata
        KEY = userdata.get('COURSE_API_KEY')
    except Exception:
        KEY = None
from openai import OpenAI
MODEL = 'qwen3.8-max'
EXTRA = {'enable_thinking': False}
client = OpenAI(api_key=KEY or 'course-key-not-configured',
    base_url='https://token-plan.ap-southeast-1.maas.aliyuncs.com/compatible-mode/v1',
    timeout=35.0, max_retries=0)
spec=importlib.util.spec_from_file_location('course_workbench',course_file('notebooks/course_workbench.py'))
workbench=importlib.util.module_from_spec(spec)
spec.loader.exec_module(workbench)
print('Course interface ready.' if KEY else 'Add COURSE_API_KEY in Colab Secrets, enable notebook access, and rerun this setup. Source browsing remains available.')
```

```python
#@title Show complete traces and judgment controls
TRACE_DATA=json.loads(course_file('data/agent-traces.json').read_text(encoding='utf-8'))
marks=workbench.show_trace_lab(TRACE_DATA,client,MODEL,EXTRA)
```

## Review the four core cases

1. Read trace 2 and the review standard. Identify the completed business action.
2. Read trace 5 in full, including its final note. Decide whether the policy
   permits routing after a disclosed search failure. Record missing policy if needed.
3. Read trace 7. Separate routing completion from answering the customer's question.
4. Read trace 8 and its original instructions. Compare the fallback destination
   with the valid queues. Identify the conflicting specification.
5. Save your verdict and evidence before asking the model judge.

Does a tool error make the entire run fail, or can recovery be acceptable? The
answer depends on the agreed task and fallback rules. A warning flag identifies
something to inspect; it does not replace the acceptance standard.

## Compare judgments

1. Ask the model judge about each marked trace.
2. Identify whether a disagreement concerns a fact, interpretation, or missing rule.
3. Amend the review standard in the text field and judge again.
4. Record the reason for changing the standard and download your judgments.

The live judge may agree with you. Agreement is not proof, and disagreement is
not automatically a model error. Read the cited evidence. A model judge is a
model call in this notebook; it does not operate an agent tool loop.
